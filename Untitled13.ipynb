```python
from langchain_core.documents import Document

# Create Langchain documents

doc1 = Document(
    page_content="Virat Kohli is one of the most successful and consistent batsmen in cricket history. He has captained Royal Challengers Bengaluru in the IPL.",
    metadata={
        "player": "Virat Kohli",
        "team": "RCB",
        "role": "Batsman"
    }
)

doc2 = Document(
    page_content="Rohit Sharma is known for his elegant batting and leadership skills. He is the captain of Mumbai Indians and one of India's top opening batsmen.",
    metadata={
        "player": "Rohit Sharma",
        "team": "Mumbai Indians",
        "role": "Batsman"
    }
)

doc3 = Document(
    page_content="MS Dhoni is one of the greatest captains and wicketkeepers in cricket history. He plays for Chennai Super Kings in the IPL.",
    metadata={
        "player": "MS Dhoni",
        "team": "Chennai Super Kings",
        "role": "Wicketkeeper"
    }
)

doc4 = Document(
    page_content="Jasprit Bumrah is a world-class fast bowler famous for his yorkers and unique bowling action. He plays for Mumbai Indians.",
    metadata={
        "player": "Jasprit Bumrah",
        "team": "Mumbai Indians",
        "role": "Bowler"
    }
)

doc5 = Document(
    page_content="Ravindra Jadeja is an excellent all-rounder known for his batting, bowling, and fielding skills. He represents Chennai Super Kings in the IPL.",
    metadata={
        "player": "Ravindra Jadeja",
        "team": "Chennai Super Kings",
        "role": "All-Rounder"
    }
)

```

```python
docs=[doc1,doc2,doc3,doc4,doc5]
```

```python
!pip install -qU langchain-huggingface sentence-transformers chromadb
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma

embedding = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

vector_store = Chroma(
    embedding_function=embedding,
    persist_directory="chroma_db",
    collection_name="players"
)

vector_store.add_documents(docs)

print("Documents Added Successfully")
```

```python
vector_store.get(include=['embeddings','documents','metadatas'])
```

```python
vector_store.similarity_search(query="who is greatest batsman of all time ?",k=2)
```

```python
vector_store.similarity_search(query="who is greatest batsman of all time ?",k=2,filter={"team":"Chennai Super Kings"})
```
