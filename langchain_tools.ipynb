how to make a custom tools in langchain


```python
# step 1 create a function
def multiply(a,b):
  """/multiply two numbers"""
  return a*b
```

step 2 add type hints


```python
def multiply(a:int, b:int)->int:
  """Multiply two number"""
  return a*b
```

add tool decorator


```python
from langchain_core.tools import tool
```

```python
@tool
def multiply(a:int,b:int)->int:
  """Multiply two number"""
  return a*b
```

```python
result=multiply.invoke({"a":5,"b":3})
```

```python
print(result)
```

incustom method there are 3 types by which u can create a tool
1 by decorator
2 by pysantic model
3 by basetool method

```python
from langchain.tools import BaseTool
```

```python
from typing import Type
from pydantic import BaseModel, Field
```

```python
class MultiplyInput(BaseModel):
    a: int = Field(description="First number")
    b: int = Field(description="Second number")
```

```python
class MultiplyTool(BaseTool):
  name: str="multiply"
  description: str="multyply two numbers "

  args_schema:Type[BaseModel]=MultiplyInput

  def _run(self,a:int, b:int)->int:
    return a*b
```

```python
multiply_tool=MultiplyTool()
```

```python
res=multiply_tool.invoke({'a':8,'b':3})
```

```python
print(res)
```
